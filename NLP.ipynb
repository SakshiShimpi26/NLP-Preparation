## **Stemming**

**PorterStemmer**

```python
from nltk.stem import PorterStemmer
```

```python
ps = PorterStemmer()

words = ["running", "hopping", "caresses", "ponies", "relational", "hopefulness", "digitizer", "activation", "control"]

for word in words:
    stemmed = ps.stem(word)
    print(word+" ==> "+stemmed)
```

## **Lemmatization**

```python
import nltk
from nltk.stem import WordNetLemmatizer
```

```python
words = ["running", "hopping", "caresses", "ponies", "relational", "hopefulness", "digitizer", "activation", "control"]

lemmatizer = WordNetLemmatizer()

# If POS(Part Of Speech is not provided it gives the lemma as noun)
for word in words:
    lemma = lemmatizer.lemmatize(word)
    print(word+" ==> "+lemma)
```

```python
# POS here is v ==> verb
print(lemmatizer.lemmatize("running","v"))
```

```python
# POS here is a ==> adjective
print(lemmatizer.lemmatize("better", "a") )
```

```python
lemmatizer.lemmatize("better", "v")
```

## **Stop-word Removing**

```python
from nltk.corpus import stopwords
nltk.download('stopwords')
```

```python
stop_words = stopwords.words('english')
print(stop_words)
```

```python
example = """Natural language processing (NLP) is a subfield of computer science and artificial intelligence (AI)
          that uses machine learning to enable computers to understand and communicate with human language."""

cleaned_by_removing_stopwords = []
words = nltk.word_tokenize(example)
for word in words:
  if word.lower() not in stop_words:
    cleaned_by_removing_stopwords.append(word)

print(cleaned_by_removing_stopwords)
```

## **Bag of Words (BOW)**

```python
from sklearn.feature_extraction.text import CountVectorizer
```

```python
corpus = ["He is a good Boy", "She is a good Girl", "Boy and Girl are good"]
```

```python
# Data Preprocessing and cleaning

l1 = []
for i in range(0,len(corpus)):
  l1.append(corpus[i].lower())

l2 = []
stop_word = stopwords.words('english')
for i in range(0,len(l1)):
  x = l1[i].split(" ")
  s=[]
  for a in x:
    if a not in stop_word:
      s.append(a)
  l2.append(" ".join(s))
print(l2)
```

```python
vectorizer = CountVectorizer()
X=vectorizer.fit_transform(l2).toarray()
print(vectorizer.get_feature_names_out())
X
```

## **TF-IDF (Term Frequency - Inverse Document Frequency)**

```python
from sklearn.feature_extraction.text import TfidfVectorizer
```

```python
print(l2)
```

```python
vectorizer = TfidfVectorizer()
tfidf_matrix = vectorizer.fit_transform(l2)
```

```python
print(tfidf_matrix.toarray())
```

```python
import pandas as pd
# Convert to DataFrame for better readability
df = pd.DataFrame(tfidf_matrix.toarray(), columns=vectorizer.get_feature_names_out())
```
